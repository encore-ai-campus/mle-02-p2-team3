# Q/A JSONL 기록 구성 점검
합성 그래프와 합성 질문으로 레코드를 만들며 DB나 LLM에 연결하지 않습니다.

```python
from pathlib import Path
import sys
CHAT_DIR = Path.cwd() / 'src/card_benefit/payment_card/file_pipeline/chat'
OUTPUT_PATH = Path('/tmp/fp-qa-log-check.jsonl')
sys.path.insert(0, str(CHAT_DIR))
import qa_log

synthetic_result = {'answer': '월 5만원', 'selected_entities': ['합성 카드'], 'subgraph': {'nodes': [{'id': 'node-1', 'name': '합성 카드', 'props': {'한도': '5만원', 'run_id': 'run-demo'}}], 'edges': []}}
record = qa_log.build_record(question='합성 한도 질문', answer='월 5만원', result=synthetic_result, latency_ms={'total': 2.1}, source_refs=[{'node_id': 'node-1', 'doc_sha256': 'a' * 64, 'page_no': 1, 'start_pos': 2, 'end_pos': 8, 'quote_status': 'located'}])
print({key: record[key] for key in ('answer_format_version', 'retrieved_node_count', 'source_refs', 'format_check')})
print('unguarded number candidates:', record['ungrounded_numbers'])
print('JSONL write:', qa_log.write_record(record, OUTPUT_PATH))
```
